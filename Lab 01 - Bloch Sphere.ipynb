## Interactive Bloch Sphere

This program provides an interactive visualization of a single-qubit state on the Bloch sphere. The qubit state is parameterized by the polar angle $\theta$ and the azimuthal angle $\varphi$:

$$
|S\rangle=
\cos\left(\frac{\theta}{2}\right)|0\rangle+
e^{i\varphi}\sin\left(\frac{\theta}{2}\right)|1\rangle,
\qquad
0\le\theta\le\pi,\quad
0\le\varphi<2\pi.
$$

The sliders allow the user to change the values of $\theta$ and $\varphi$ interactively. For each pair of angles, the program displays the corresponding qubit state, the measurement probabilities, and the position of the Bloch vector on the Bloch sphere.

To draw the Bloch vector, the spherical coordinates $(\theta,\varphi)$ are converted into Cartesian coordinates according to

$$
x=\sin\theta\cos\varphi,\qquad
y=\sin\theta\sin\varphi,\qquad
z=\cos\theta.
$$

These coordinates define the unit Bloch vector

$$
\mathbf{r}=(x,y,z),
$$

which uniquely represents the pure qubit state on the surface of the Bloch sphere.

According to the Born rule, the probabilities of obtaining the measurement outcomes $0$ and $1$ in the computational basis are given by the squared magnitudes of the corresponding probability amplitudes:

$$
P(0)=\left|\cos\left(\frac{\theta}{2}\right)\right|^2
=\cos^2\left(\frac{\theta}{2}\right),
$$

$$
P(1)=\left|e^{i\varphi}\sin\left(\frac{\theta}{2}\right)\right|^2
=\sin^2\left(\frac{\theta}{2}\right).
$$

Since $|e^{i\varphi}|=1$, the azimuthal angle $\varphi$ does not affect the measurement probabilities in the computational basis. Instead, it determines the orientation of the Bloch vector around the $z$-axis, while the polar angle $\theta$ determines the probability distribution between the measurement outcomes $0$ and $1$.

**NOTE:** The program requires the `ipywidgets` package to create and display the interactive sliders in the Jupyter Notebook.



```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

from fractions import Fraction
from IPython.display import display, Math
from qiskit.visualization import plot_bloch_vector


def pi_fraction(angle, max_denominator=16):
    """
    Convert an angle in radians to an approximate fraction
    or multiple of π.
    """

    fraction = Fraction(
        angle / np.pi
    ).limit_denominator(max_denominator)

    numerator = fraction.numerator
    denominator = fraction.denominator

    if numerator == 0:
        return "0"

    if denominator == 1:
        if numerator == 1:
            return r"\pi"
        return rf"{numerator}\pi"

    if numerator == 1:
        return rf"\frac{{\pi}}{{{denominator}}}"

    return rf"\frac{{{numerator}\pi}}{{{denominator}}}"


@widgets.interact(
    theta=widgets.FloatSlider(
        value=np.pi / 2,
        min=0,
        max=np.pi,
        step=0.01,
        description="θ:",
        continuous_update=True,
        readout_format=".3f",
        layout=widgets.Layout(width="400px")
    ),

    phi=widgets.FloatSlider(
        value=0,
        min=0,
        max=2 * np.pi,
        step=0.01,
        description="φ:",
        continuous_update=True,
        readout_format=".3f",
        layout=widgets.Layout(width="400px")
    )
)
def bloch_simulator(theta, phi):

    # --------------------------------------------------------
    # Probability amplitudes
    # --------------------------------------------------------

    a = np.cos(theta / 2)
    b = np.exp(1j * phi) * np.sin(theta / 2)

    # --------------------------------------------------------
    # Bloch-vector coordinates
    # --------------------------------------------------------

    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)

    # --------------------------------------------------------
    # Born probabilities
    # --------------------------------------------------------

    p0 = abs(a) ** 2
    p1 = abs(b) ** 2

    # --------------------------------------------------------
    # Display angles
    # --------------------------------------------------------

    display(
        Math(
            rf"\theta={theta:.3f}\ \mathrm{{rad}}"
            rf"\ \left({pi_fraction(theta)}\right),"
            rf"\qquad"
            rf"\varphi={phi:.3f}\ \mathrm{{rad}}"
            rf"\ \left({pi_fraction(phi)}\right)"
        )
    )

    # --------------------------------------------------------
    # Display the qubit state
    # --------------------------------------------------------

    display(
        Math(
            rf"|S\rangle="
            rf"{a:.3f}|0\rangle"
            rf"+"
            rf"\left({b.real:.3f}{b.imag:+.3f}i\right)|1\rangle"
        )
    )

    # --------------------------------------------------------
    # Display Bloch-vector coordinates
    # --------------------------------------------------------

    display(
        Math(
            rf"\mathbf{{r}}="
            rf"({x:.3f},\,{y:.3f},\,{z:.3f})"
        )
    )

    # --------------------------------------------------------
    # Display Born probabilities
    # --------------------------------------------------------

    display(
        Math(
            rf"P(0)={p0:.3f},"
            rf"\qquad "
            rf"P(1)={p1:.3f}"
        )
    )

    # --------------------------------------------------------
    # Display the Bloch sphere
    # --------------------------------------------------------

    figure = plot_bloch_vector(
        [x, y, z],
        title="Qubit state"
    )

    figure.set_size_inches(3.5, 3.5)

    display(figure)
    plt.close(figure)
```
